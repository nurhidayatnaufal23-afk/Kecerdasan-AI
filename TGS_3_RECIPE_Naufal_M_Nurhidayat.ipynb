# BAGIAN I — RECIPE DARI BUKU

## 1. Recipe Pemeriksaan Kualitas Data Stasiun dalam CSV

### Permasalahan
Buat alur kerja untuk mengevaluasi sebuah file CSV yang berisi data stasiun. Pemeriksaan berfokus pada susunan kolom, periode waktu, data kosong, dan waktu pencatatan yang berulang. Tujuan akhirnya adalah laporan pemeriksaan kualitas data, bukan penafsiran kondisi cuaca. File asli harus tetap utuh.

### Teori Singkat
Pemeriksaan kualitas data atau QA/QC dilakukan sebelum data dipakai dalam analisis. Kesalahan format waktu, nilai kosong, dan catatan ganda dapat menyebabkan hasil perhitungan menjadi tidak tepat. Karena itu, kondisi data perlu diringkas terlebih dahulu.

### Masukan dan Batasan
- `CSV_PATH`: lokasi file CSV yang akan diperiksa.
- `TIME_COLUMN`: nama kolom berisi waktu pengamatan.
- `UNITS`: keterangan satuan pada kolom yang digunakan.
- File dibuka hanya untuk dibaca; data mentah tidak ditimpa.
- Nilai yang hilang atau ganda dilaporkan, bukan dihapus secara otomatis.
- Kesimpulan dibatasi pada pemeriksaan yang benar-benar dilakukan.

### Prompt dan Tahapan
> Rancang recipe Python untuk menilai kualitas sebuah CSV stasiun. Tinjau kolom, format waktu, rentang pengamatan, nilai kosong, serta timestamp yang berulang. Sajikan hasil pemeriksaan secara ringkas tanpa memodifikasi file sumber. Bedakan rancangan program dari hasil eksekusi yang telah dibuktikan.

Urutan kerja:
1. Membaca file CSV secara aman.
2. Memeriksa daftar kolom dan kolom waktu.
3. Mengubah kolom waktu ke format tanggal-waktu untuk keperluan pemeriksaan.
4. Menghitung nilai kosong dan timestamp yang berulang.
5. Menyusun ringkasan jumlah baris, kolom, dan rentang waktu.
6. Menuliskan masalah yang ditemukan beserta status pemeriksaannya.

### Rencana Implementasi
Pustaka `pandas` dapat digunakan untuk membaca tabel dan membuat ringkasan, sedangkan `pathlib` dapat membantu mengelola lokasi file. Contoh lokasi kode: `code/recipes/station_csv_check.py`. Lokasi tersebut masih berupa rancangan jika file belum dibuat.

### Output yang Diharapkan
- `station-check-report.md`: ringkasan pemeriksaan.
- `station-check-record.json`: catatan parameter dan status proses jika program dijalankan.

Laporan berisi dimensi data, nama kolom, periode waktu, jumlah nilai kosong, jumlah timestamp berulang, serta catatan kolom yang tidak sesuai.

### Pemeriksaan Tambahan
Pastikan jumlah baris pada sumber tidak berubah, laporan terisi, dan status proses ditulis secara jujur. Gunakan status `Not yet executed` apabila program belum benar-benar dijalankan.

## 2. Recipe Validasi dan Penataan Data GeoJSON

### Permasalahan
Susun prosedur untuk memeriksa file GeoJSON, lalu menghasilkan salinan yang telah ditata berdasarkan aturan yang jelas, daftar fitur dalam CSV, dan ringkasan pemeriksaan berbentuk JSON. Pemeriksaan mencakup struktur, geometri, atribut, koordinat, serta informasi sistem koordinat.

### Teori Singkat
GeoJSON menyimpan objek geografis dalam struktur JSON. Sebuah file perlu diperiksa dari sisi sintaks, tipe objek, geometri, koordinat, dan properti setiap fitur. Pada koordinat geografis GeoJSON, susunan koordinat umumnya ditulis sebagai bujur kemudian lintang. Informasi sistem koordinat tidak boleh ditebak apabila tidak tercantum atau tidak diketahui.

### Masukan dan Batasan
- `GEOJSON_PATH`: alamat file GeoJSON sumber.
- `OUTPUT_GEOJSON`: lokasi salinan keluaran.
- `OUTPUT_CSV`: lokasi tabel inventaris fitur.
- `OUTPUT_JSON`: lokasi ringkasan validasi.
- Fitur dan atribut tidak boleh dihapus tanpa aturan yang disepakati.
- Geometri kosong atau tidak valid harus ditandai.
- Berhasil membaca JSON belum tentu berarti seluruh geometri valid.

### Prompt dan Tahapan
> Buat recipe untuk menginspeksi GeoJSON dan menyiapkan keluaran GeoJSON, CSV inventaris, serta JSON ringkasan. Periksa struktur file, tipe geometri, atribut, koordinat, geometri kosong atau bermasalah, dan informasi CRS. Jangan mengubah koordinat atau menyatakan data valid tanpa pemeriksaan yang mendukung.

Langkah kerja:
1. Membuka file dan memeriksa validitas sintaks JSON.
2. Memastikan struktur utama dan daftar fitur tersedia.
3. Mencatat tipe geometri dan atribut pada setiap fitur.
4. Memeriksa koordinat dan geometri kosong.
5. Menandai informasi CRS yang tidak tersedia atau belum pasti.
6. Menata keluaran hanya berdasarkan aturan yang ditetapkan.
7. Mengekspor inventaris fitur dan ringkasan pemeriksaan.
8. Mencatat fitur yang gagal diproses.

### Rencana Implementasi
Modul `json` dapat dipakai untuk membaca struktur dasar, sedangkan `geopandas` atau pustaka geospasial lain dapat digunakan apabila tersedia untuk memeriksa fitur dan geometri. Dependensi yang digunakan perlu dicatat.

### Output yang Diharapkan
- GeoJSON hasil penataan.
- CSV berisi inventaris fitur.
- JSON berisi ringkasan QA/QC.
- Catatan proses dan daftar peringatan jika ada.

### Pemeriksaan Tambahan
Bandingkan jumlah fitur sebelum dan sesudah proses, periksa atribut penting, dan validasi kembali file keluaran. Jika kode belum dijalankan, keluaran tersebut tetap disebut sebagai target, bukan hasil yang sudah tersedia.

## 3. Recipe Simulasi Difusi Panas pada Satu Dimensi

### Permasalahan
Buat prosedur simulasi yang memperkirakan perubahan suhu sepanjang suatu garis terhadap waktu. Model memerlukan profil suhu awal, difusivitas termal, kondisi batas, jarak antar titik grid, dan interval waktu.

### Teori Singkat
Persamaan difusi panas satu dimensi dinyatakan sebagai:

\[
\frac{\partial T}{\partial t}=\alpha\frac{\partial^2T}{\partial x^2}
\]

Keterangan:
- \(T\): suhu.
- \(t\): waktu.
- \(x\): posisi.
- \(\alpha\): difusivitas termal.

Untuk metode beda hingga eksplisit standar, nilai parameter kestabilan dapat ditulis:

\[
r=\frac{\alpha\Delta t}{(\Delta x)^2}
\]

Pada skema eksplisit satu dimensi yang umum, kestabilan mensyaratkan \(r\leq \frac{1}{2}\). Ketentuan ini perlu diperiksa kembali sesuai metode numerik yang dipilih.

### Masukan dan Batasan
- `INITIAL_PROFILE`: nilai suhu pada kondisi awal.
- `ALPHA`: difusivitas termal beserta satuannya.
- `BOUNDARIES`: kondisi batas model.
- `DX`: jarak antar titik ruang.
- `DT`: interval waktu simulasi.
- Seluruh besaran harus memakai satuan yang konsisten.
- Hasil simulasi tidak boleh disebut konvergen sebelum diuji.

### Prompt dan Tahapan
> Rancang recipe simulasi difusi panas 1D berdasarkan profil awal, difusivitas, kondisi batas, jarak grid, dan langkah waktu yang diberikan. Jelaskan persamaan model, validasi satuan, pemeriksaan kestabilan, pembanding sederhana, visualisasi, serta pencatatan status proses. Jangan menyatakan simulasi telah berhasil jika belum dijalankan.

Langkah kerja:
1. Memeriksa parameter dan satuan masukan.
2. Membentuk grid ruang serta grid waktu.
3. Menetapkan suhu awal dan kondisi batas.
4. Menghitung nilai parameter kestabilan.
5. Melakukan iterasi numerik jika program telah tersedia.
6. Membandingkan hasil dengan kasus sederhana atau solusi analitik bila memungkinkan.
7. Menyimpan data suhu, grafik, dan catatan proses.

### Rencana Implementasi
`numpy` dapat digunakan untuk operasi numerik dan `matplotlib` untuk membuat grafik profil suhu. Metode yang dipilih, nilai parameter, dan versi pustaka sebaiknya dicatat.

### Output yang Diharapkan
- Tabel suhu berdasarkan posisi dan waktu.
- Grafik perubahan suhu.
- Ringkasan parameter simulasi.
- Catatan eksekusi yang memuat status dan peringatan.

### Pemeriksaan Tambahan
Pastikan kondisi batas diterapkan dengan benar, satuan konsisten, dan hasil tidak menunjukkan perilaku numerik yang tidak wajar. Klaim kalibrasi atau konvergensi memerlukan pengujian tersendiri.

## 4. Recipe Pengolahan Data Magnetik untuk Pemetaan Anomali

### Permasalahan
Merancang alur pengolahan data survei magnetik dari pemeriksaan data lapangan hingga visualisasi anomali magnetik. Prosedur ditujukan agar data dapat ditelusuri, parameter pengolahan tercatat, dan data mentah tidak tertimpa selama proses.

### Teori Singkat
Metode magnetik memanfaatkan variasi medan magnet bumi yang dipengaruhi oleh sifat kemagnetan material bawah permukaan. Data lapangan dapat mengandung pengaruh medan utama bumi, variasi harian, gangguan lokal, dan kesalahan pengukuran. Pengolahan biasanya memerlukan pemeriksaan kualitas, koreksi yang sesuai dengan metode survei, serta pemetaan nilai anomali. Pilihan koreksi harus mengikuti jenis data dan informasi yang benar-benar tersedia.

Secara konseptual, anomali magnetik dapat ditulis:

\[
\Delta T = T_{\text{observasi}}-T_{\text{rujukan}}
\]

dengan \(T_{\text{observasi}}\) sebagai medan magnet terukur dan \(T_{\text{rujukan}}\) sebagai nilai acuan yang ditentukan oleh prosedur pengolahan. Persamaan ini merupakan gambaran umum; penentuan nilai acuan dan koreksi harus mengikuti metode yang digunakan.

### Masukan dan Batasan
Masukan utama:
- `MAGNETIC_DATA_PATH`: lokasi file data survei.
- `LATITUDE` dan `LONGITUDE`: koordinat titik pengukuran.
- `MAGNETIC_FIELD`: nilai medan magnet terukur.
- `DATETIME`: waktu pengukuran apabila tersedia.
- `STATION_ID`: identitas titik pengukuran apabila tersedia.
- `REFERENCE_FIELD`: data acuan yang dipakai dalam pengolahan, jika tersedia.

Batasan:
1. Data asli dibaca tanpa ditimpa.
2. Satuan medan magnet harus diperiksa, misalnya nT.
3. Koordinat dan waktu harus ditinjau sebelum pemetaan.
4. Data kosong dan pengukuran berulang dilaporkan.
5. Koreksi hanya diterapkan jika informasi dan metode pendukung tersedia.
6. Peta anomali tidak langsung dianggap sebagai bukti jenis batuan atau struktur tertentu.

### Prompt dan Tahapan
**Prompt yang dapat digunakan kembali:**

> Bertindaklah sebagai asisten pengolahan data geofisika metode magnetik. Periksa file `[MAGNETIC_DATA_PATH]` tanpa mengubah data mentah. Validasi kolom, koordinat, waktu pengukuran, satuan, nilai kosong, dan data berulang. Jika data dan parameter pendukung memadai, hitung nilai anomali sesuai metode yang ditetapkan, buat tabel keluaran dan peta, serta dokumentasikan asumsi, koreksi, dan status eksekusi. Jangan membuat interpretasi geologi yang tidak didukung oleh data.

**Urutan prosedur:**
1. Membaca file dan mencatat jumlah baris serta kolom.
2. Memastikan kolom koordinat dan medan magnet tersedia.
3. Memeriksa format koordinat, satuan, nilai kosong, dan duplikasi.
4. Meninjau apakah data acuan atau koreksi yang dibutuhkan tersedia.
5. Menghitung anomali hanya setelah metode dan nilai acuan ditetapkan.
6. Membuat visualisasi sebaran titik atau peta kontur jika distribusi data mendukung.
7. Menyimpan tabel hasil, gambar, dan metadata pengolahan.
8. Mencatat kendala dan status pelaksanaan.

### Rencana Implementasi
Pustaka yang dapat dipertimbangkan:
- `pandas` untuk membaca dan memeriksa tabel.
- `numpy` untuk operasi numerik.
- `matplotlib` untuk visualisasi.
- `pathlib` dan `json` untuk pengelolaan file serta catatan proses.

Contoh struktur direktori:

```text
magnetic-survey/
├── data/
│   └── magnetic_observation.csv
├── scripts/
│   └── magnetic_processing.py
└── results/
    ├── magnetic_report.md
    ├── anomaly_map.png
    ├── anomaly_values.csv
    └── process_record.json
```

Struktur ini merupakan usulan penyimpanan dan bukan bukti bahwa program telah dieksekusi.

### Output yang Diharapkan
- `magnetic_report.md`: laporan pemeriksaan dan metode.
- `anomaly_map.png`: peta hasil jika data memadai.
- `anomaly_values.csv`: tabel nilai hasil pengolahan.
- `process_record.json`: parameter, asumsi, dan status proses.

### Pemeriksaan Tambahan
- [ ] File masukan terbaca dan tetap tidak berubah.
- [ ] Kolom wajib dan satuan telah diperiksa.
- [ ] Koordinat berada pada format yang sesuai.
- [ ] Nilai kosong dan pengukuran berulang telah dilaporkan.
- [ ] Metode koreksi dan data acuan tercatat.
- [ ] Peta dilengkapi legenda, skala, dan sistem koordinat jika relevan.
- [ ] Status proses sesuai dengan bukti yang tersedia.

Status yang dapat digunakan: `Draft`, `Not yet executed`, `Partial`, atau `Executed`.

### Keterbatasan dan Risiko
Kualitas hasil dipengaruhi oleh ketelitian instrumen, kondisi survei, jarak antarstasiun, variasi waktu, dan ketepatan koreksi. Peta anomali magnetik perlu ditafsirkan bersama informasi geologi dan data pendukung lainnya. Tanpa informasi koreksi yang cukup, hasil hanya boleh dilaporkan sebagai pemeriksaan atau visualisasi awal.

### Ringkasan Recipe Baru
Recipe ini menyediakan langkah kerja yang tertata untuk mengevaluasi data magnetik, memeriksa kelengkapan informasi, menghitung anomali ketika syaratnya terpenuhi, serta menyajikan hasil dalam bentuk tabel dan peta. Dokumentasi parameter dan keterbatasan membantu menjaga agar proses dapat diulang dan hasilnya tidak ditafsirkan secara berlebihan.
